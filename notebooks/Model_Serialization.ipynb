Import Libraries

```python
import pandas as pd
import numpy as np
import pickle
import os

from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor

from statsmodels.tsa.arima.model import ARIMA
```

Load Dataset

```python
df = pd.read_csv("../data/processed/feature_engineering_data.csv")

df.head()
```

Metals

```python
metals = [
    "Price_gold",
    "Price_silver",
    "Price_alum",
    "Price_nickel",
    "Price_uran"
]
```

Save Linear Regression model

```python
for metal in metals:

    X = df[
        [
            "Year",
            "Month_Num",
            "Quarter",
            f"{metal}_Lag_1",
            f"{metal}_Lag_2",
            f"{metal}_Lag_3",
            f"{metal}_Rolling_Mean_3"
        ]
    ]

    y = df[metal]

    model = LinearRegression()

    model.fit(X, y)

    file_name = (
        metal.replace("Price_", "")
        + "_lr.pkl"
    )

    with open(
        f"../models/linear_regression/{file_name}",
        "wb"
    ) as f:

        pickle.dump(model, f)

    print(f"Saved: {file_name}")
```

Save Random Forest model

```python
for metal in metals:

    X = df[
        [
            "Year",
            "Month_Num",
            "Quarter",
            f"{metal}_Lag_1",
            f"{metal}_Lag_2",
            f"{metal}_Lag_3",
            f"{metal}_Rolling_Mean_3"
        ]
    ]

    y = df[metal]

    model = RandomForestRegressor(
        n_estimators=100,
        random_state=42
    )

    model.fit(X, y)

    file_name = (
        metal.replace("Price_", "")
        + "_rf.pkl"
    )

    with open(
        f"../models/random_forest/{file_name}",
        "wb"
    ) as f:

        pickle.dump(model, f)

    print(f"Saved: {file_name}")
```

Save XGBoost model

```python
for metal in metals:

    X = df[
        [
            "Year",
            "Month_Num",
            "Quarter",
            f"{metal}_Lag_1",
            f"{metal}_Lag_2",
            f"{metal}_Lag_3",
            f"{metal}_Rolling_Mean_3"
        ]
    ]

    y = df[metal]

    model = XGBRegressor(
        n_estimators=100,
        learning_rate=0.1,
        max_depth=5,
        random_state=42
    )

    model.fit(X, y)

    file_name = (
        metal.replace("Price_", "")
        + "_xgb.pkl"
    )

    with open(
        f"../models/xgboost/{file_name}",
        "wb"
    ) as f:

        pickle.dump(model, f)

    print(f"Saved: {file_name}")
```

Save ARIMA model

```python
for metal in metals:

    print(f"Training ARIMA for {metal}")

    series = df[metal]

    model = ARIMA(
        series,
        order=(5,1,0)
    )

    fitted_model = model.fit()

    file_name = (
        metal.replace("Price_", "")
        + "_arima.pkl"
    )

    with open(
        f"../models/arima/{file_name}",
        "wb"
    ) as f:

        pickle.dump(
            fitted_model,
            f
        )

    print(f"Saved: {file_name}")
```

Check whether the models saved are loaded or not

```python
model_paths = [
    "../models/linear_regression",
    "../models/random_forest",
    "../models/xgboost",
    "../models/arima"
]

for folder in model_paths:

    print(f"\nChecking {folder}")

    for file in os.listdir(folder):

        path = os.path.join(folder, file)

        with open(path, "rb") as f:

            model = pickle.load(f)

        print(f"Loaded: {file}")
```
